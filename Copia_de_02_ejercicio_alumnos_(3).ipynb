### 5. Análisis de impacto: decisiones

* **`pais`**
* **Qué significa que falte:** Errores de tipeo, espacios en blanco o falta de registro al momento de capturar el pedido.
* **Decisión tomada:** **Imputar (reemplazar por `NaN` y luego imputar con la moda o categoría constante)**.
* **Por qué:** Los nulos representan solo un 3.33% (10 registros). Eliminar filas perdería información de las transacciones, por lo que es preferible unificar los textos vacíos y asignar la moda (país más frecuente) o el valor `"Desconocido"`.


* **`monto`**
* **Qué significa que falte:** Error en el registro de la transacción o fallo en la integración del sistema de cobro.
* **Decisión tomada:** **Imputar (mediana)** *(o **eliminar filas**, según prefieras)*.
* **Por qué:** El monto es un dato clave para el análisis financiero. Al ser pocos nulos (9 registros), imputar con la mediana no distorsiona la distribución monetaria ni los totales generales, evitando descartar las ventas.


* **`edad`**
* **Qué significa que falte:** El cliente prefirió no proporcionar su edad durante el registro o la compra.
* **Decisión tomada:** **Conservar los NaN** *(o **imputar por la mediana**)*.
* **Por qué:** La edad es un dato demográfico opcional. Conservar los nulos evita alterar la distribución real de edades del público o sesgar análisis segmentados por rango etario.


* **`calificacion`**
* **Qué significa que falte:** El cliente realizó la compra pero decidió no valorar la experiencia o el producto.
* **Decisión tomada:** **Conservar los NaN**.
* **Por qué:** No haber calificado es una respuesta en sí misma. Imputar una nota promedio alteraría la satisfacción real medida de los usuarios que sí decidieron opinar.


* **`cupon`**
* **Qué significa que falte:** La compra se realizó a precio de lista, sin aplicar ningún código de descuento.
* **Decisión tomada:** **Imputar (constante `"Sin Cupón"`)** o **Conservar los NaN**.
* **Por qué:** La ausencia de cupón refleja una condición normal del negocio. Reemplazar los nulos por la etiqueta `"Sin Cupón"` facilita el filtrado y agrupamiento en análisis posteriores.


* **`telefono_alternativo`**
* **Qué significa que falte:** Información de contacto secundaria opcional que la mayoría de los usuarios no proporciona.
* **Decisión tomada:** **Conservar los NaN** *(o **eliminar la columna** si no se usará en el análisis)*.
* **Por qué:** Al faltar en la gran mayoría de los registros (248 nulos), no aporta valor predictivo o analítico significativo, pero conservarla con nulos no interfiere con el resto de los datos.

## 6. Aplicar la limpieza
Implementá tus decisiones sobre una copia (`df_limpio`). Orden sugerido: duplicados → columna → filas → imputaciones.

```python
# ✏️ df_limpio = ...
# 0. Crear la copia para trabajar de forma segura
df_limpio = df_pedidos.copy()

# ---------------------------------------------------------
# 1. DUPLICADOS
# ---------------------------------------------------------
# Eliminar registros duplicados por id_pedido conservando la primera ocurrencia
df_limpio = df_limpio.drop_duplicates(subset=["id_pedido"], keep="first")


# ---------------------------------------------------------
# 2. COLUMNAS
# ---------------------------------------------------------
# (Opcional) Si decides eliminar 'telefono_alternativo' por exceso de nulos:
df_limpio = df_limpio.drop(columns=["telefono_alternativo"])


# ---------------------------------------------------------
# 3. CONVERSIÓN DE TIPOS DE DATOS Y NORMALIZACIÓN DE FILAS
# ---------------------------------------------------------
# Reemplazar textos vacíos/espacios en 'pais' por NaN
import numpy as np
df_limpio["pais"] = df_limpio["pais"].replace(r"^\s*$", np.nan, regex=True)

# Convertir tipos de datos
df_limpio["fecha_pedido"] = pd.to_datetime(df_limpio["fecha_pedido"])
df_limpio["edad"] = df_limpio["edad"].astype("Int64")
df_limpio["calificacion"] = df_limpio["calificacion"].astype("Int64")


# ---------------------------------------------------------
# 4. IMPUTACIONES
# ---------------------------------------------------------
# Pais: imputar con la moda (país más frecuente)
moda_pais = df_limpio["pais"].mode()[0]
df_limpio["pais"] = df_limpio["pais"].fillna(moda_pais)

# Monto: imputar con la mediana
mediana_monto = df_limpio["monto"].median()
df_limpio["monto"] = df_limpio["monto"].fillna(mediana_monto)

# Cupon: imputar nulos con la etiqueta 'Sin Cupón'
df_limpio["cupon"] = df_limpio["cupon"].fillna("Sin Cupón")

# Nota: 'edad' y 'calificacion' se conservan con NaN según las decisiones tomadas.


# ---------------------------------------------------------
# VERIFICACIÓN FINAL
# ---------------------------------------------------------
df_limpio.info()
```
