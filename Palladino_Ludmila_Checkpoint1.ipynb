# Saneamiento y Selección.

Pregunta de investigacion:
¿Qué factores demográficos, de salud y de comunicación se relacionan con la inasistencia a los turnos médicos?



```python

# Primero Filtramos pacientes con edades entre 0 y 100 años.
df_filtrado = df[(df['Age'] >= 0) & (df['Age'] <= 100)]
```

```python
# Verificar las dimensiones del nuevo dataset
print("Dimensiones originales:", df.shape)
print("Dimensiones filtradas:", df_filtrado.shape)
```

```python
# Mostrar las primeras filas
display(df_filtrado.head())
```

¿Qué hace este código?
- df['Age'] >= 0: selecciona registros con edades iguales o mayores a cero.
- df['Age'] <= 100: selecciona registros con edades de hasta 100 años.
- &: combina ambas condiciones; tienen que cumplirse las dos.

El resultado se guarda en df_filtrado, por lo que conservás el dataset original en df.

```python
# Seleccionar pacientes adultos que no asistieron
df_inasistencias = df_filtrado[
    (df_filtrado['Age'] >= 18) &
    (df_filtrado['No-show'] == 'Yes')
]
```

```python
# Mostrar resultados
display(df_inasistencias.head())
```

```python
print("Total de turnos de adultos con inasistencia:",
      df_inasistencias.shape[0])
```

En este dataset, No-show == 'Yes' indica que el paciente no asistió al turno, mientras que No-show == 'No' indica que sí asistió.

**Eliminar columnas innecesarias.**

Para analizar los factores asociados a la asistencia, los identificadores individuales (id)  no suelen aportar información explicativa directa. Al eliminarlos conservarmos las variables relevantes.

```python

# Eliminar columnas identificadoras
df_analisis = df_filtrado.drop(
    columns=['PatientId', 'AppointmentID']
)
```

```python
# Verificar el resultado
print("Dimensiones del dataset seleccionado:", df_analisis.shape)

display(df_analisis.head())
```

Se aplicaron filtros booleanos para realizar una limpieza del conjunto de datos y obtener subconjuntos relevantes para la investigación sobre los factores asociados a la inasistencia a turnos médicos.

En primer lugar, se filtraron los registros para conservar pacientes con edades entre 0 a 100 años, con el objetivo de excluir valores fuera del rango definido para el análisis. Posteriormente, se creó un subconjunto de pacientes adultos que no asistieron a sus turnos, utilizando las variables Age y No-show.

Además, se eliminaron las columnas PatientId y AppointmentID, dado que funcionan como identificadores y no aportan información explicativa.
Se conservaron las variables demográficas, de salud, ubicación, programación de turnos y recepción de mensajes, junto con No-show como variable objetivo.

Estas operaciones permiten organizar los datos y concentrar el análisis en las variables que podrían estar relacionadas con la asistencia o inasistencia de los pacientes.
